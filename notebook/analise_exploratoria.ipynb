## **3 Tratamento dos Dados**

### 3.1 Tratamento de tipos incorretos

```python
# Conversão das colunas 'DataInicial' e 'DataFinal' para datetime
df['DataInicial'] = pd.to_datetime(df['DataInicial'], format='%d/%m/%Y')
print("Conversão da coluna 'DataInicial' para datetime feita com sucesso!")

df['DataFinal'] = pd.to_datetime(df['DataFinal'], format='%d/%m/%Y')
print("\nConversão da coluna 'DataFinal' para datetime feita com sucesso!")
```

### 3.2 Criação das colunas 'Total' e 'Média'

```python
# Criação das colunas 'Total' e 'Média'
colunas_doencas = ['Pneumonias', 'Asma', 'Bronquite', 'Bronquiolite',
                   'IVAS', 'Influenza', 'IRA', 'Sinusite']
df['TotalIDR'] = df[colunas_doencas].sum(axis=1)
df['MediaIDR'] = df[colunas_doencas].mean(axis=1)

# Imprimir informações
print("Colunas 'Total' e 'Média' criadas com sucesso!")
df[['TotalIDR', 'MediaIDR']].head()
```

### 3.3 Criação de DataFrame Simplificado

```python
# Criação de DataFrame Simplificado
df_simplificado = df[['DataInicial', 'DataFinal', 'FocosDeQueimadas', 'TotalIDR', 'MediaIDR']].copy()
print("DataFrame simplificado criado com sucesso!")

# Exibindo o DataFrame Simplificado
print("\nExibindo as cinco primeiras linhas do DataFrame simplificado")
df_simplificado.head()
```
